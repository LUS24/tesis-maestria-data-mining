```python
from pathlib import Path
from docling.document_converter import DocumentConverter
```

```python
# Obtener Código de Minería
BASE_PATH = Path.cwd()
BASE_FILE_PATH = BASE_PATH.joinpath("codigo_minero_43797.html")
ENCODING = "ISO-8859-1"

source = "http://servicios.infoleg.gob.ar/infolegInternet/anexos/40000-44999/43797/texact.htm"  # document per local path or URL
converter = DocumentConverter()
result = converter.convert(source)

with open(BASE_FILE_PATH.with_suffix(".md"), "w", encoding="utf-8") as f:
    f.write(result.document.export_to_markdown())

with open(BASE_FILE_PATH.with_suffix(".html"), "w", encoding="utf-8") as f:
    f.write(result.document.export_to_html())
```

Extract file components.

Notes:
- Use logic to determine the type of regulation.
- ULID: Universally Unique Lexicographically Sortable Identifier

- main_regulation:
    - ulid
        - data_type: ULID
        - example: "01ARZ3NDEKTSV4RRFFQ69G5FAV"
    - regulation_type:
        - data_type: str
        - example: "law", "code", "resolution", "decree", "disposition"
    - identifier:
        - data_type: str
        - example: "2408/25", "11179"
    - title:
        - data_type: str
        - example: "código penal de la nación argentina"
    - summary: str
        - data_type: str
        - example: "Apruébase el código penal."
    - sanction_date:
        - data_type: date "YYYY-MM-DD"
        - example: 1921-09-30
    - sanction_formula:
        - data_type: str
        - example: "El Senado y la Cámara de Diputados, reunidos en Congreso, sancionan con fuerza de LEY"
    - sanction_method:
        - data_type: str
        - example: ""
    - sanction_authority:
        - data_type: str
        - example: "National Congress", ""
    - promulgation_date:
        - data_type: date "YYYY-MM-DD"
        - example: 1921-09-30
    - publication_date:
        - data_type: date "YYYY-MM-DD"
    - status:
        - data_type: str
        - example: "in force", "derogated", "repealed", "suspended", "invalid"
    - modified_by: list[main_regulation-reference]
    - signed_by: 
        - data_type: list[person-object]
        - example: [
            {
                "names": "Carlos Saúl"
                "last_name": "Menem"
                "full_name": "Carlos Saúl Menem"
                "position": "President"
            },
        ]

    - parts:
        - data_type: list[regulation_part-object]
        - example: [
            {
                "": ""
                "": ""
                "": ""
                "": ""
            },
        ]

- person:
    - ulid
        - data_type: ULID
        - example: "01ARZ3NDEKTSV4RRFFQ69G5FAV"
    - first_name
        - data_type: str
        - example: "Carlos"
    - last_name
        - data_type: str
        - example: "Menem"
    - full_name
        - data_type: str
        - example: "Carlos Saúl Menem"


- regulation_part:
    - ulid
        - data_type: ULID
        - example: "01ARZ3NDEKTSV4RRFFQ69G5FAV"
    - parent_regulation: ULID
    - part_type: str (header, )
