```python
from pydantic_ai import Agent, RunContext
from dotenv import load_dotenv
from pydantic_ai import Agent

import nest_asyncio
nest_asyncio.apply() # To be able to run pydantic in jupyter notebooks preventing event loop issues


load_dotenv()

translator_agent = Agent(  
    'openai:gpt-5',
    # output_type=str,
    system_prompt=(
        'I will give you a word in english, please translate it to Spanish, for example:'
        "cake: torta"
    ),
)

result = translator_agent.run_sync("code")

```
