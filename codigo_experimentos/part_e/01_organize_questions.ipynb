# Organize questions

Organize questions to have 1 question per line with its respective response.

```python
import json
from pathlib import Path

BASE_PATH = Path.cwd()
DATA_FILE_PATH = BASE_PATH.joinpath("20250804_101435_generated_questions - 130825.json")
OUTPUT_FILE_PATH = BASE_PATH.joinpath("20251012_llm_answers_baseline_from_20250804_101435_generated_questions.json")


with open(DATA_FILE_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
tipo_preguntas = set()

for pregunta in data["preguntas"]:
    tipo_preguntas.add(pregunta["tipo_pregunta"])

print(tipo_preguntas)
```

```python
preguntas = []

# Extract text from questions and store in a unified format
for pregunta in data["preguntas"]:
    if pregunta["tipo_pregunta"] == "teaching_multiple_many_questions_many_answers":
        texto = pregunta.get("texto_enunciado")
        for sub_pregunta in pregunta["sub_preguntas"]:
            preguntas.append(
                {
                    "numero_pregunta": pregunta.get("numero_pregunta"),
                    "numero_subpregunta": sub_pregunta.get("numero_subpregunta"),
                    "texto": texto + " " +sub_pregunta.get("texto_subpregunta"),
                    "contexto_para_respuesta": sub_pregunta.get("contexto_para_respuesta"),
                    "respuesta": sub_pregunta.get("respuesta"),
                }
            )
    elif pregunta["tipo_pregunta"] == "teaching_multiple_single_question_single_answer":
        preguntas.append(
            {
                "numero_pregunta": pregunta.get("numero_pregunta"),
                "numero_subpregunta": None,
                "texto": pregunta.get("texto_enunciado") + " "+pregunta.get("texto_pregunta"),
                "contexto_para_respuesta": pregunta.get("contexto_para_respuesta"),
                "respuesta": pregunta.get("respuesta"),
            }
        )
    else:
        raise ValueError("Improperly formatted question")

```

```python
with open(OUTPUT_FILE_PATH, encoding="utf-8", mode="w") as f:
    f.write(json.dumps(preguntas, ensure_ascii=False, indent=4))
```
