```python
from pathlib import Path
import json
import copy

BASE_PATH = Path.cwd().resolve()
INPUT_PATH = BASE_PATH.joinpath("llm_responses_agentic_rag_results_FINAL.json")
OUTPUT_PATH = BASE_PATH.joinpath("llm_responses_agentic_rag_results_FINAL_fixed.json")

with open(INPUT_PATH, mode="r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
missing_responses = []

for i, pregunta in enumerate(data):
    for respuesta in pregunta["llm_responses_agentic_rag"]:
        if not respuesta["texto"]:
            missing_responses.append(
                (i, respuesta)
            )
```
