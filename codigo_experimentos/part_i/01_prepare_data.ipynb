```python
import polars as pl
from pathlib import Path
import json

BASE_PATH = Path.cwd().parent
SOURCE_DATA_PATH = BASE_PATH.joinpath("part_h", "llm_responses_all_results_with_metrics_FINAL.json")
with open(SOURCE_DATA_PATH, mode="r", encoding="utf-8") as f:
    source_data = json.load(f)
```

## Prepare Text Ouput Data

This will allow calculation of extra metrics if required

```python
OUTPUT_PATH = BASE_PATH.joinpath("part_i", "responses_text_cleaned.csv")

cols = [
    "nro_pregunta",
    "nro_subpregunta",
    "texto_pregunta",
    "respuesta_esperada",
    "estrategia",
    "id_modelo_especialista",
    "respuesta_obtenida",
]
estrategias = [
    "llm_responses_no_rag",
    "llm_responses_base_rag",
    "llm_responses_agentic_rag",
]

observations = []

for datapoint in source_data:
    nro_pregunta = datapoint["numero_pregunta"]
    nro_subpregunta = datapoint["numero_subpregunta"]
    texto_pregunta = datapoint["texto"]
    respuesta_esperada = datapoint["respuesta"]
    for estrategia in estrategias:
        for data_estrategia  in datapoint[estrategia]:
            if estrategia == "llm_responses_agentic_rag":
                id_modelo_especialista = data_estrategia["law_specialist_model"].split(":")[1]
            else:
                id_modelo_especialista = data_estrategia["metadata"]["usage"]["cost"]["model_id"]
            respuesta_obtenida = data_estrategia["texto"]
            observations.append([nro_pregunta, nro_subpregunta, texto_pregunta, respuesta_esperada, estrategia, id_modelo_especialista, respuesta_obtenida])

pl.DataFrame(observations, schema=cols, orient="row").write_csv(OUTPUT_PATH, separator=",")
```

## Prepare Metrics data

```python
OUTPUT_PATH = BASE_PATH.joinpath("part_i", "metric_comparison_source_data_cleaned.csv")

cols = [
    "nro_pregunta",
    "nro_subpregunta",
    "estrategia",
    "id_modelo_especialista",
    "metrica",
    "valor",
]
estrategias = [
    "llm_responses_no_rag",
    "llm_responses_base_rag",
    "llm_responses_agentic_rag",
]

observations = []

for datapoint in source_data:
    nro_pregunta = datapoint["numero_pregunta"]
    nro_subpregunta = datapoint["numero_subpregunta"]
    for estrategia in estrategias:
        for data_estrategia  in datapoint[estrategia]:
            if estrategia == "llm_responses_agentic_rag":
                id_modelo_especialista = data_estrategia["law_specialist_model"].split(":")[1]
            else:
                id_modelo_especialista = data_estrategia["metadata"]["usage"]["cost"]["model_id"]
            for metrica, valor in data_estrategia["metrics"].items():
                valor = round(valor["value"], 4)
                observations.append([nro_pregunta, nro_subpregunta, estrategia, id_modelo_especialista, metrica, valor])

pl.DataFrame(observations, schema=cols, orient="row").write_csv(OUTPUT_PATH, separator=",")
```
